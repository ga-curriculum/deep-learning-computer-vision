### 1. What is a Deep Neural Network?

**Definition:**
A Deep Neural Network (DNN) is a neural network with multiple layers between the input and output layers. It consists of:
- **Input Layer:** Accepts input features.
- **Hidden Layers:** Perform computations and transformations.
- **Output Layer:** Provides predictions.

### 2. Define a Simple Neural Network

**Step 1: Define the Model Structure**
In PyTorch, we define a neural network by subclassing `torch.nn.Module`.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# Define a simple DNN
class SimpleDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(SimpleDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# Initialize the model
input_size = 2
hidden_size = 4
output_size = 1
model = SimpleDNN(input_size, hidden_size, output_size)
print(model)
```

**Exercise 1: Modify Model Structure**
1. Add a new hidden layer with 6 neurons to the `SimpleDNN` class.

**Solution:**

```python
# Solution for Exercise 1
class UpdatedDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(UpdatedDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.fc3 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x

# Initialize the updated model
model = UpdatedDNN(input_size=2, hidden_size=6, output_size=1)
print(model)
```

```python
class MultiLayerDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(MultiLayerDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.fc3 = nn.Linear(hidden_size, hidden_size)
        self.fc4 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = F.relu(self.fc3(x))
        x = self.fc4(x)
        return x

# Initialize the new model
model = MultiLayerDNN(input_size, hidden_size, output_size)
print(model)
```

### 3. Forward Pass and Activation Functions

**Forward Pass:**
The forward pass refers to the process of passing input data through the network to produce an output.

```python
# Example data
inputs = torch.tensor([[1.0, 2.0], [3.0, 4.0]])

# Perform a forward pass
outputs = model(inputs)
print("Outputs:", outputs)
```

```python
# Activation Function Demo
x = torch.linspace(-5, 5, 100)
relu = F.relu(x)
sigmoid = torch.sigmoid(x)
tanh = torch.tanh(x)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(x.numpy(), relu.numpy(), label='ReLU')
plt.plot(x.numpy(), sigmoid.numpy(), label='Sigmoid')
plt.plot(x.numpy(), tanh.numpy(), label='Tanh')
plt.legend()
plt.title("Activation Functions")
plt.show()
```

**Exercise 2: Experiment with Activation Functions**
Modify the `SimpleDNN` to use different activation functions for each hidden layer.

**Solution:**

```python
class CustomActivationDNN(nn.Module):
    def __init__(self, input_size, hidden_size, output_size):
        super(CustomActivationDNN, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.fc3 = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        x = torch.sigmoid(self.fc1(x))
        x = F.tanh(self.fc2(x))
        x = self.fc3(x)
        return x

# Initialize the new model
model = CustomActivationDNN(input_size, hidden_size, output_size)
print(model)
```
