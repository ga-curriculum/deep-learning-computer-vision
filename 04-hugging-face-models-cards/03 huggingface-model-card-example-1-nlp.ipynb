# Sentiment Analysis with Hugging Face

In this exercise, you will:
- Load a pre-trained model for sentiment analysis from Hugging Face.
- Use the model to analyze the sentiment of sample text inputs.
- Explore and interpret the model's outputs.

## Step 1: Install and Import Required Libraries
We'll use the `transformers` library from Hugging Face to load the model.

```python
# Install Hugging Face Transformers library (uncomment if needed)
# !pip install transformers

# Import libraries
from transformers import pipeline
```

## Step 2: Load the Pre-Trained Model
We'll load a sentiment analysis pipeline using the `transformers` library.

```python
# Load the sentiment-analysis pipeline
sentiment_pipeline = pipeline("sentiment-analysis")

# Test the pipeline
print("Pipeline loaded successfully!")
```

## Step 3: Analyze Sentiment of Sample Text
Input a few example sentences to see how the model performs.

```python
# Define some sample sentences
sample_texts = [
    "I love using Hugging Face models! They're so easy to use.",
    "This is the worst product I've ever used.",
    "The weather today is quite pleasant, isn't it?"
]

# Analyze sentiment
results = sentiment_pipeline(sample_texts)

# Display results
for text, result in zip(sample_texts, results):
    print(f"Text: {text}")
    print(f"Sentiment: {result['label']} (Confidence: {result['score']:.2f})")
    print()
```

- Write your own sentences and analyze their sentiment using the model.

```python
# Your turn: Input custom sentences
custom_texts = [
    "Add your own sentence here!",
    "Try another one!"
]

# Analyze custom sentences
custom_results = sentiment_pipeline(custom_texts)

# Display results
for text, result in zip(custom_texts, custom_results):
    print(f"Text: {text}")
    print(f"Sentiment: {result['label']} (Confidence: {result['score']:.2f})")
    print()
```
