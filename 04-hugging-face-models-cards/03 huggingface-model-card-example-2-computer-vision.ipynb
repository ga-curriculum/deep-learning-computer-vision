## Step 1: Install and Import Required Libraries
We'll use the `transformers` library from Hugging Face and `Pillow` for image handling.

```python
# Install required libraries (uncomment if needed)
# !pip install transformers pillow

# Import libraries
from transformers import pipeline
from PIL import Image
import requests
```

## Step 2: Load the Pre-Trained Model
We'll load an image classification pipeline using a pre-trained ResNet model.

```python
# Load the image classification pipeline
image_classifier = pipeline("image-classification")

# Test the pipeline
print("Pipeline loaded successfully!")
```

## Step 3: Classify an Example Image
We'll use an example image from the web to test the model.

```python
# Load an example image
image_url = "https://upload.wikimedia.org/wikipedia/commons/thumb/a/a8/Golden_Retriever_Carlos_%2810581910556%29.jpg/800px-Golden_Retriever_Carlos_%2810581910556%29.jpg"
image = Image.open(requests.get(image_url, stream=True).raw)

# Display the image
image.show()

# Classify the image
results = image_classifier(image)

# Display results
for result in results:
    print(f"Label: {result['label']}, Confidence: {result['score']:.2f}")
```
